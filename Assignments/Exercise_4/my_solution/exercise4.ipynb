```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import rotate
import math

# Load CIFAR-10 dataset
def load_cifar10_sample():
    """
    Load a small sample from CIFAR-10 for experiments
    Return: tensor of shape (100, 3, 32, 32)
    """
    # TODO: Implement this function
    transform = transforms.ToTensor()
    cifar10 = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)

    sample_images = torch.stack([cifar10[i][0] for i in range(100)])

    return sample_images

# Implement rotation function
def rotate_image(image, angle, mode='constant', cval=0):
    """
    Rotate image by given angle
    Args:
        image: tensor of shape (C, H, W)
        angle: rotation angle in degrees
        mode: padding mode ('constant', 'reflect', 'wrap')
        cval: constant value for padding
    Returns:
        rotated image of same shape
    """
    # TODO: Implement rotation with different padding modes
    # Hint: Use scipy.ndimage.rotate or implement your own
    
    if isinstance(image, torch.Tensor):
        image = image.cpu().numpy()

    C, H, W = image.shape
    rotated = []

    for c in range(C):
        rot = rotate(
            image[c], angle, reshape=True, order=1, mode=mode, cval=cval
        )  # rotated shape > original
        rotated.append(rot)

    rotated = np.stack(rotated, axis=0)  # (C, H', W')

    # --- Center crop back to (H, W) ---
    new_H, new_W = rotated.shape[1:]
    top = (new_H - H) // 2
    left = (new_W - W) // 2
    cropped = rotated[:, top:top+H, left:left+W]

    return torch.tensor(cropped, dtype=torch.float32)


    return transformed_img
```

```python
def analyze_rotation_artifacts(image, angles=[0, 45, 90, 135, 180, 225, 270, 315]):
    """
    Analyze artifacts introduced by rotation with different padding strategies
    
    Tasks:
    1. Rotate the image using 'constant' padding (zero-padding)
    2. Rotate the image using 'reflect' padding
    3. Rotate the image using 'wrap' padding
    4. Visualize the differences
    5. Calculate pixel-wise error between original and rotate-back images
    
    Return: Dictionary with errors for each padding mode
    """
    # TODO: Implement the analysis
    modes = ['constant', 'reflect', 'wrap']
    errors = {mode: [] for mode in modes}
    fig, axs = plt.subplots(len(angles), len(modes), figsize=(len(modes) * 4, len(angles) * 3))
    
    for i, angle in enumerate(angles):
        for j, mode in enumerate(modes):
            rotated = rotate_image(image, angle, mode=mode)
            rotated_back = rotate_image(rotated, -angle, mode=mode)

            # Calculate mean absolute error
            error = torch.mean(torch.abs(image - rotated_back))
            errors[mode].append(error)

            # Visualization
            img = (image - rotated_back).clamp(0, 1).permute(1, 2, 0).cpu().numpy()
            axs[i, j].imshow((img * 255).astype(np.uint8))
            axs[i, j].set_title(f"{mode}, angle={angle}")
            axs[i, j].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    return errors
```

```python
def quantitative_rotation_analysis(images, conv_layer):
    """
    Quantitatively analyze how rotation affects CNN features
    
    Args:
        images: batch of images (N, C, H, W)
        conv_layer: a Conv2d layer with 7x7 kernel
    
    Tasks:
    1. Apply convolution to original images
    2. For each rotation angle in [0, 45, 90, ..., 315]:
       a. Rotate input image
       b. Apply convolution
       c. Rotate feature map back
       d. Calculate MSE with original feature map
    3. Compare errors with and without padding
    4. Plot error as function of rotation angle
    
    Return: Dictionary with error metrics
    """
    # TODO: Implement the analysis
    
    # Define your 7x7 convolutional layer here
    conv_7x7 = nn.Conv2d(3, 64, kernel_size=7, padding=3)
    
    errors_with_padding = {}
    errors_without_padding = {}
    
    # Your implementation here
    angles=[0, 45, 90, 135, 180, 225, 270, 315]

    fig, axs = plt.subplots(len(angles), figsize=(2, 4))

    original_features = conv_7x7(images)
    
    for i, angle in enumerate(angles):
        rotated = rotate_image(images, angle)
        conv_img = conv_7x7(rotated)
        rotated_back = rotate_image(conv_img, -angle)

        # Calculate mean absolute error
        mse = F.mse_loss(rotated_back, original_features).item()
        errors_without_padding[angle] = mse


        rotated = rotate_image(images, angle, cval=10)
        conv_img = conv_7x7(rotated)
        rotated_back = rotate_image(conv_img, -angle)

        # Calculate mean absolute error
        mse = F.mse_loss(rotated_back, original_features).item()
        errors_with_padding[angle] = mse




    # Plotting
    plt.figure(figsize=(8, 4))
    plt.plot(angles, [errors_without_padding[a] for a in angles], label='Without Padding', marker='o')
    plt.plot(angles, [errors_with_padding[a] for a in angles], label='With Reflect Padding', marker='o')
    plt.xlabel("Rotation Angle (degrees)")
    plt.ylabel("MSE with Original Features")
    plt.title("Rotation Artifact Analysis")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    return {
        'with_padding': errors_with_padding,
        'without_padding': errors_without_padding
    }
```

```python
class GroupConvC4(nn.Module):
    """
    Group Convolution layer equivariant to C4 group (90 degree rotations)
    """
    
    def __init__(self, in_channels, out_channels, kernel_size=3):
        super().__init__()
        self.in_channels = in_channels
        self.out_channels = out_channels
        self.kernel_size = kernel_size
        
        # TODO: Initialize parameters
        # Hint: You need kernels for each group element
        self.base_kernel = nn.Parameter(
            torch.randn(out_channels, in_channels, kernel_size, kernel_size)
        )

    def _get_rotation_matrices(self):
        """
        Return the 4 rotation matrices for C4 group
        Returns: List of 2x2 rotation matrices
        """
        # TODO: Implement
        # Return matrices for 0°, 90°, 180°, 270° rotations
        def _rot_gen(angle):
            angle = angle * np.pi / 180
            mat = torch.tensor([
                [math.cos(angle), -math.sin(angle)],
                [math.sin(angle), math.cos(angle)]
            ])

            return mat 
        
        return [_rot_gen(0), _rot_gen(90), _rot_gen(180), _rot_gen(270)]
    
    def _rotate_kernel(self, kernel, angle):
        """
        Rotate a convolution kernel by given angle
        Args:
            kernel: tensor of shape (out_channels, in_channels, H, W)
            angle: rotation angle (0, 90, 180, or 270)
        """
        # TODO: Implement kernel rotation
        # Hint: You need to rotate and permute dimensions correctly
        assert angle in [0, 90, 180, 270]
        k = angle // 90

        return torch.rot90(kernel, k=k, dims=[-2, -1])

    
    def lifting_convolution(self, x):
        """
        Lift a function on R^2 to a function on SE(2)
        Args:
            x: input tensor (B, C, H, W)
        Returns:
            lifted tensor (B, C, 4, H, W) - 4 for C4 group
        """
        # TODO: Implement lifting operation
        # Apply convolution with rotated versions of the kernel
        batch_size = x.shape[0]
        
        # Initialize output
        lifted = torch.zeros(batch_size, self.out_channels, 4, 
                           x.shape[2], x.shape[3])
        
        # For each rotation in C4
        for i, angle in enumerate([0, 90, 180, 270]):
            # Your implementation here
            rotated_kernel = self._rotate_kernel(self.base_kernel, angle)
            lifted[:, :, i, :, :] = F.conv2d(x, rotated_kernel, padding=self.kernel_size // 2)
            
        return lifted
    
    def group_convolution(self, lifted_x):
        """
        Perform group convolution on lifted feature maps
        Args:
            lifted_x: tensor of shape (B, C, 4, H, W)
        Returns:
            tensor of shape (B, C, 4, H, W)
        """
        # TODO: Implement group convolution
        # This should respect the group structure
        
        batch_size, channels, group_size, H, W = lifted_x.shape
        output = torch.zeros_like(lifted_x)
        
        # For each output group element
        for g in range(group_size):
            # For each input group element
            out_g = torch.zeros(
                batch_size, self.out_channels, H, W, device=lifted_x.device, dtype=lifted_x.dtype
            )
            for h in range(group_size):
                # Calculate g * h^(-1) to determine which kernel to use
                # Your implementation here
                g_inv = (g - h) % group_size # g * h^-1 mod 4
                rotated_kernel = self._rotate_kernel(self.base_kernel, g_inv * 90)
                x_h = lifted_x[:, :, h, :, :]
                out_g += F.conv2d(x_h, rotated_kernel, padding=self.kernel_size // 2)
            output.append(out_g)

        return output
    
    def forward(self, x, lifted_input_type=False): # Renamed arg for clarity
        if not lifted_input_type: # Input is (B, self.in_channels, H, W)
            return self.lifting_convolution(x) # Output (B, self.out_channels, 4, H,W)
        else: # Input is (B, self.in_channels, 4, H, W)
            return self.group_convolution(x) # Output (B, self.out_channels, 4, H,W)
```
